```python
import pandas as pd

# Import the pandas library so we can work with tabular data in a DataFrame.
# Read the Nobel Prize winners CSV file into a DataFrame named `df`.
# This DataFrame will store all rows and columns from the dataset for analysis.
df = pd.read_csv("nobel_prize_winners.csv")
```

```python
# Count how often each unique row appears in the temporary DataFrame.
# The DataFrame.value_counts() method groups rows by all column values
# and returns the frequency of each row combination.
temp.value_counts()
```

```python
# Count non-missing genders in the Nobel dataset.
gender_counts = df['gender'].dropna().value_counts()

# Count non-missing birth countries in the Nobel dataset.
birth_country_counts = df['birth_country'].dropna().value_counts()

# The first index in the value_counts Series is the most frequent category.
# This works because value_counts sorts counts in descending order by default.
top_gender = gender_counts.index[0]

# The most frequent birth country is also the first index in that counts Series.
top_country = birth_country_counts.index[0]

# Print a concise summary of the most common gender and birth country,
# including the number of awards associated with each category.
print(f'Most commonly awarded gender: {top_gender} ({gender_counts.iloc[0]} awards)')
print(f'Most commonly awarded birth country: {top_country} ({birth_country_counts.iloc[0]} awards)')
```

```python
# Create a decade column from awardYear
df["decade"] = (df["awardYear"] // 10) * 10

# Total Nobel Prize winners by decade
total_by_decade = df.groupby("decade").size()

# US-born Nobel Prize winners by decade
usa_winners_by_decade = df[df["birth_country"].eq("USA")].groupby("decade").size()

# Ratio of US-born winners to all winners by decade
ratio_by_decade = usa_winners_by_decade / total_by_decade

# Choose the decade with the highest ratio
max_decade_usa = int(ratio_by_decade.idxmax())

print(max_decade_usa)
```

```python
# Compute female laureate proportion by decade and Nobel Prize category
decade_category_totals = df.groupby(["decade", "category"]).size()
female_counts = df[df["gender"].eq("female")].groupby(["decade", "category"]).size()

female_proportions = female_counts / decade_category_totals

# Find the decade/category combination with the highest female share
best_pair = female_proportions.idxmax()

# Store as requested: decade is the key, category is the value
max_female_dict = {int(best_pair[0]): best_pair[1]}
max_female_dict
```

```python
# Identify the earliest female Nobel laureate in the dataset
female_laureates = df[df["gender"].eq("female")].dropna(subset=["awardYear"])
first_female_row = female_laureates.sort_values("awardYear").iloc[0]

first_woman_name = str(first_female_row["name"])      # "Marie Curie"
first_woman_category = str(first_female_row["category"])  # "Physics"

first_woman_name, first_woman_category
```

```python
# Identify individuals or organizations that won more than one Nobel Prize
repeat_counts = df.groupby("id").size()
repeat_ids = repeat_counts[repeat_counts > 1].index

repeat_list = []

for laureate_id in repeat_ids:
    laureate_rows = df[df["id"] == laureate_id]
    
    # Prefer the full human name when available
    if laureate_rows["fullName"].notna().any():
        winner_name = laureate_rows["fullName"].dropna().iloc[0]
    # Otherwise use organization name
    elif laureate_rows["orgName"].notna().any():
        winner_name = laureate_rows["orgName"].dropna().iloc[0]
    # Finally fall back to the name field
    else:
        winner_name = laureate_rows["name"].dropna().iloc[0]
    
    repeat_list.append(winner_name)

repeat_list
```
